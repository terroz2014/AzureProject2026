```python
%sql
CREATE TABLE IF NOT EXISTS project_dev.gold.dim_patient
(
    patient_key STRING,
    src_patientid STRING,
    firstname STRING,
    lastname STRING,
    middlename STRING,
    ssn STRING,
    phonenumber STRING,
    gender STRING,
    dob DATE,
    address STRING,
    datasource STRING
);
```

```python
%sql
TRUNCATE TABLE project_dev.gold.dim_patient;
```

```python
%sql
INSERT INTO project_dev.gold.dim_patient
SELECT
    patient_key,
    src_patientid,
    firstname,
    lastname,
    middlename,
    ssn,
    phonenumber,
    gender,
    dob,
    address,
    datasource
FROM project_dev.silver.patients
WHERE is_current = true
  AND is_quarantined = false;
```
