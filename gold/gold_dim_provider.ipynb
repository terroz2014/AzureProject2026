```python
%sql
INSERT INTO project_dev.gold.dim_provider
SELECT
    ProviderID,
    FirstName,
    LastName,
    concat(DeptID, '-', datasource) AS deptid,
    NPI,
    datasource
FROM project_dev.silver.providers
WHERE is_quarantined = false;
```
