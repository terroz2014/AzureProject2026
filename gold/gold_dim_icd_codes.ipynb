```python

%sql
INSERT INTO project_dev.gold.dim_icd
select distinct
  icd_code,
  icd_code_type,
  code_description,
  current_timestamp() refreshed_at
from
  project_dev.silver.icd_codes
where
  is_current_flag = true
```
