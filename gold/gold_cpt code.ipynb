```python
%sql
INSERT INTO project_dev.gold.dim_cpt_code
SELECT
    cpt_codes,
    procedure_code_category,
    procedure_code_descriptions,
    code_status,
    current_timestamp() AS refreshed_at
FROM project_dev.silver.cptcodes
WHERE is_quarantined = false
  AND is_current = true;
```
